```python
import os
os.chdir(r'C:\Users\bashe\Downloads\asd')
import pandas as pd
import numpy as np
import pickle
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LogisticRegression

# ── Load CSV ──────────────────────────────────────────────────────
df = pd.read_csv(r'C:\Users\bashe\Downloads\train.csv')

# ── Clean ─────────────────────────────────────────────────────────
df.replace(["?", " ", ""], pd.NA, inplace=True)

# ── Standardise gender to readable labels: 'Male' / 'Female' ─────
df['gender'] = df['gender'].astype(str).str.strip().str.lower()
df['gender'] = df['gender'].map({'m': 'Male', 'f': 'Female'})

# ── Standardise ethnicity — capitalise properly ───────────────────
df['ethnicity'] = df['ethnicity'].astype(str).str.strip().str.title()

# ── Standardise jaundice & austim ────────────────────────────────
df['jaundice'] = df['jaundice'].astype(str).str.strip().str.lower().map({'yes': 1, 'no': 0})
df['austim']   = df['austim'].astype(str).str.strip().str.lower().map({'yes': 1, 'no': 0})

# ── Drop rows with missing values in key columns ─────────────────
df = df.dropna(subset=['age', 'gender', 'ethnicity', 'jaundice', 'austim', 'Class/ASD'])

# ── Fit encoders on clean string labels ──────────────────────────
gender_encoder    = LabelEncoder()
ethnicity_encoder = LabelEncoder()

df['gender_enc']    = gender_encoder.fit_transform(df['gender'])
df['ethnicity_enc'] = ethnicity_encoder.fit_transform(df['ethnicity'])

# ── Print exact classes (copy these into app.py selectbox) ────────
print(" Gender classes    :", list(gender_encoder.classes_))
print("Ethnicity classes :", list(ethnicity_encoder.classes_))

# ── Build feature matrix (5 features only) ───────────────────────
X = df[['age', 'gender_enc', 'ethnicity_enc', 'jaundice', 'austim']]
y = df['Class/ASD']

print(" Training shape:", X.shape)

# ── Train fresh model on EXACTLY 5 features ──────────────────────
model = LogisticRegression(max_iter=1000)
model.fit(X, y)

# ── Save all 3 pkl files ─────────────────────────────────────────
pickle.dump(model,             open('model.pkl',             'wb'))
pickle.dump(gender_encoder,    open('gender_encoder.pkl',    'wb'))
pickle.dump(ethnicity_encoder, open('ethnicity_encoder.pkl', 'wb'))

print(" model.pkl saved — expects:", model.n_features_in_, "features")
print(" gender_encoder.pkl saved")
print(" ethnicity_encoder.pkl saved")
```
